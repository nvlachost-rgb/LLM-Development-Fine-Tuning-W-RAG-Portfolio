```python
import json
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
from google.colab import drive


drive.mount('/content/drive')


db_path = "/content/drive/MyDrive/LLM Development Portfolio/Datasets/JSON Datasets/erp_master_database.json"
with open(db_path, "r", encoding="utf-8") as f:
    master_data = json.load(f)


texts = []
metadatas = []


for client in master_data.get("clients", []):

    search_text = f"Πελάτης: {client.get('name', '')}, Τοποθεσία: {client.get('city', '')}"
    texts.append(search_text)

    metadatas.append({
        "type": "client",
        "name": client.get("name"),
        "afm": client.get("afm"),
        "postal_code": client.get("postal_code")
    })


for product in master_data.get("products", []):
    search_text = f"Προϊόν: {product.get('product_name', '')}, Κατηγορία: {product.get('category', '')}"
    texts.append(search_text)
    metadatas.append({
        "type": "product",
        "product_name": product.get("product_name"),
        "sku": product.get("sku"),
        "barcode": product.get("barcode"),
        "unit_price": product.get("unit_price")
    })

print(f"Ετοιμάστηκαν {len(texts)} εγγραφές για διανυσματοποίηση (Embeddings)...")


embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")


vector_db = FAISS.from_texts(texts, embeddings, metadatas=metadatas)

faiss_save_path = "/content/drive/MyDrive/erp_faiss_index"
vector_db.save_local(faiss_save_path)

print(f"Η βάση RAG δημιουργήθηκε επιτυχώς! Αποθηκεύτηκε στο: {faiss_save_path}")
```
