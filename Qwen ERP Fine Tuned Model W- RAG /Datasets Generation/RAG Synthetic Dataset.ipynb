````python
import os
import time
import json
from google import genai
from google.colab import userdata
from google.colab import drive

drive.mount('/content/drive')

api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)

prompt = """
Παράγαγε μια ψεύτικη αλλά ρεαλιστική ελληνική επιχειρησιακή βάση δεδομένων (Master Data) για μια εταιρεία αγροτικών εφοδίων.
Πρέπει να περιέχει δύο κατηγορίες:
1. "clients": Λίστα με 20 πελάτες (αγρότες, συνεταιρισμούς). Κάθε πελάτης να έχει: "name", "afm" (9ψήφιο), "city", "postal_code" (ΤΚ), "phone".
2. "products": Λίστα με 20 προϊόντα/εφόδια (σπόροι, λιπάσματα, ζωοτροφές). Κάθε προϊόν να έχει: "sku", "product_name", "category", "barcode" (13ψήφιο), "unit_price".

ΑΥΣΤΗΡΟΣ ΚΑΝΟΝΑΣ: Επέστρεψε το αποτέλεσμα ΑΠΟΚΛΕΙΣΤΙΚΑ ως ένα έγκυρο JSON αντικείμενο (object) με τα κλειδιά "clients" και "products". Μην βάλεις καμία εισαγωγή, κανέναν επίλογο και καθόλου markdown (όχι ```json). Μόνο το raw JSON.
"""

batches_to_generate = 40
output_file = "/content/drive/MyDrive/erp_master_database.json"

master_db = {
    "clients": [],
    "products": []
}

print(f"Ξεκινάει η παραγωγή Master Database στο {output_file}...")

for i in range(batches_to_generate):
    try:
        response = client.models.generate_content(
            model='gemini-3.5-flash-lite',
            contents=prompt
        )

        clean_text = response.text.strip()
        if clean_text.startswith("```json"):
            clean_text = clean_text[7:-3].strip()
        elif clean_text.startswith("```"):
            clean_text = clean_text[3:-3].strip()

        batch_data = json.loads(clean_text)

        if "clients" in batch_data:
            master_db["clients"].extend(batch_data["clients"])
        if "products" in batch_data:
            master_db["products"].extend(batch_data["products"])

        print(f"Batch {i+1}/{batches_to_generate} ολοκληρώθηκε. Σύνολο -> Πελάτες: {len(master_db['clients'])}, Προϊόντα: {len(master_db['products'])}")
        time.sleep(3)

    except Exception as e:
        print(f"Σφάλμα στο batch {i+1}: {e}")
        time.sleep(5)

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(master_db, f, ensure_ascii=False, indent=4)

print(f"\nΗ Master Database είναι έτοιμη και αποθηκεύτηκε στο: {output_file}")
````
