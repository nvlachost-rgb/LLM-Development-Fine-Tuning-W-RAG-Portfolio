````python
import os
import time
import json
from google import genai
from google.colab import userdata
from google.colab import drive


drive.mount('/content/drive')


api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)


prompt = """
Είσαι ειδικός στα ERP συστήματα και στην εφοδιαστική αλυσίδα γεωργικών προϊόντων (λιπάσματα, ζωοτροφές, ανταλλακτικά).
Παράγαγε μια λίστα με 30 συνθετικά παραδείγματα εκπαίδευσης για ένα σύστημα αυτόματης καταχώρισης παραγγελιών.

Το κάθε παράδειγμα πρέπει να περιέχει:
1. "instruction": Οδηγία (π.χ. "Ανάλυση και δομοθέτηση παραγγελίας ERP").
2. "input": Η ακατέργαστη φράση/κείμενο που γράφει βιαστικά ο υπάλληλος (π.χ. "Βασίλης Μαλαπέρδας 10 κιλά σπόρους βαμβάκι").
3. "output": Η πλήρης, δομημένη μορφή JSON που θα έπρεπε να βγάλει το ERP.

ΑΥΣΤΗΡΟΣ ΚΑΝΟΝΑΣ: Επέστρεψε τα δεδομένα ΑΠΟΚΛΕΙΣΤΙΚΑ ως μια έγκυρη JSON λίστα (array). Μην βάλεις καμία εισαγωγή, κανέναν επίλογο και καθόλου markdown (όχι ```json). Μόνο το raw JSON array.
"""


batches_to_generate = 50
output_file = "/content/drive/MyDrive/erp_training_dataset.json"
all_data = []

print(f"Ξεκινάει η παραγωγή συνθετικών δεδομένων στο {output_file}...")

for i in range(batches_to_generate):
    try:
        response = client.models.generate_content(
            model='gemini-3.5-flash-lite',
            contents=prompt
        )


        clean_text = response.text.strip()
        if clean_text.startswith("```json"):
            clean_text = clean_text[7:-3].strip()
        elif clean_text.startswith("```"):
            clean_text = clean_text[3:-3].strip()


        batch_data = json.loads(clean_text)
        all_data.extend(batch_data)

        print(f"Batch {i+1}/{batches_to_generate} ολοκληρώθηκε επιτυχώς. Σύνολο εγγραφών: {len(all_data)}")
        time.sleep(3)
    except Exception as e:
        print(f"Σφάλμα στο batch {i+1}: {e}")
        time.sleep(5)


with open(output_file, "w", encoding="utf-8") as f:
    json.dump(all_data, f, ensure_ascii=False, indent=4)

print(f"\nΤέλεια! Το {output_file} είναι έτοιμο με {len(all_data)} εγγραφές.")
````
