```python
import random


logs_path = "/content/drive/MyDrive/Datasets/Text Datasets/corpus.txt"
wiki_path = "/content/drive/MyDrive/Datasets/Text Datasets/wikipedia_el_subset.txt"
output_path = "/content/drive/MyDrive/Datasets/Text Datasets/expert1_train_final.txt"

print("Διαβάζονται τα αρχεία...")
with open(logs_path, 'r', encoding='utf-8') as f:
    logs_lines = f.readlines()

with open(wiki_path, 'r', encoding='utf-8') as f:
    wiki_lines = f.readlines()

print(f"Βρέθηκαν {len(logs_lines)} γραμμές logs και {len(wiki_lines)} γραμμές Wikipedia.")


mixed_lines = logs_lines + wiki_lines

print("Ανακάτεμα (Shuffling) για αποφυγή catastrophic forgetting...")
random.shuffle(mixed_lines)

print("Αποθήκευση στο τελικό αρχείο...")
with open(output_path, 'w', encoding='utf-8') as f:
    f.writelines(mixed_lines)

print(f"Τέλεια! Το τελικό dataset έχει συνολικά {len(mixed_lines)} γραμμές και αποθηκεύτηκε στον φάκελο 'Text Datasets'.")
```
