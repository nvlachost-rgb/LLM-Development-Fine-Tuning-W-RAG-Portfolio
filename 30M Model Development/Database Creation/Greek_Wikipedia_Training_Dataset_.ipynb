```python
from datasets import load_dataset

print("Κατέβασμα της Ελληνικής Wikipedia...")
dataset = load_dataset("wikimedia/wikipedia", "20231101.el", split="train")

num_articles = 2500

with open("wikipedia_el_subset.txt", "w", encoding="utf-8") as f:
    for i in range(num_articles):
        text = dataset[i]["text"].replace("\n\n", "\n")
        f.write(text + "\n")

print("Το wikipedia_el_subset.txt δημιουργήθηκε με επιτυχία!")
```
