```python
import os
import time
from google import genai
from google.colab import userdata
from google.colab import drive


drive.mount('/content/drive')
```

```python



api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)


prompt = """
Γράψε 50 ρεαλιστικές καταχωρήσεις (logs, tickets, παρατηρήσεις συστήματος, περιγραφές παραστατικών)
για ένα σύστημα ERP μιας ελληνικής εταιρείας γεωργικών προϊόντων (λιπάσματα, ζωοτροφές, ανταλλακτικά).

ΑΥΣΤΗΡΟΙ ΚΑΝΟΝΕΣ:
1. Χρησιμοποίησε ΑΠΟΚΛΕΙΣΤΙΚΑ Ελληνικά και Αγγλικά. Καμία άλλη γλώσσα, ούτε καν σύμβολα άλλων γλωσσών.
2. Ανάμιξε τις δύο γλώσσες ρεαλιστικά (π.χ. "Stock update στην Αποθήκη Γ", "Error: timeout στο server", "Ακύρωση παραγγελίας από το logistics department").
3. Πρόσθεσε σκόπιμα ρεαλιστικό "θόρυβο": μερικές συντομογραφίες (π.χ. τηλ, kg, τεμ), μερικά greeklish (π.χ. "paraggelia", "lipasma"), και μικρά τυπογραφικά λάθη.
4. Μην βάλεις καμία εισαγωγή, κανέναν επίλογο και κανένα markdown. Μόνο τα raw δεδομένα.
5. Κάθε καταχώρηση πρέπει να είναι σε νέα, δική της γραμμή.
"""


batches_to_generate = 500
output_file = "/content/drive/MyDrive/corpus.txt"

print(f"Ξεκινάει η παραγωγή συνθετικών δεδομένων στο {output_file}...")

with open(output_file, "a", encoding="utf-8") as f:
    for i in range(batches_to_generate):
        try:
            response = client.models.generate_content(
                model='gemini-3.5-flash-lite',
                contents=prompt
            )
            f.write(response.text + "\n")
            print(f"Batch {i+1}/{batches_to_generate} ολοκληρώθηκε επιτυχώς.")
            time.sleep(2)
        except Exception as e:
            print(f"Σφάλμα στο batch {i+1}: {e}")
            time.sleep(5)

print("\nΤέλεια! Το corpus.txt είναι έτοιμο.")
```
